# S1bis — Auto-évaluation interactive

## Structures de données Python

Ce quiz vous aide à vérifier ce que vous savez sur les listes, les tuples et les dictionnaires. Après chaque réponse, vous recevrez une correction expliquée.

### Avant de commencer

1. Ouvrez ce notebook dans Google Colab et créez votre propre copie avec **Fichier → Enregistrer une copie dans Drive**.
2. Exécutez la cellule de code ci-dessous et répondez aux questions dans la zone de saisie.
3. Les réponses et le score restent dans la session en cours : ils ne sont pas enregistrés automatiquement et seront perdus quand la session prendra fin.

Le quiz est noté sur **11 points**. L’objectif est de repérer ce qu’il faut encore réviser, pas d’obtenir une note certificative.

```python
competences = [
    "Reconnaître les listes, tuples et dictionnaires",
    "Décomposer un tuple dans des variables",
    "Ajouter une information dans un dictionnaire",
    "Choisir la structure adaptée à une situation",
]


def demander_choix(question, choix):
    """Affiche une question à choix et vérifie que la réponse est valide."""
    print("\n" + question)
    for lettre, texte in choix.items():
        print(f"  {lettre}) {texte}")

    reponses_possibles = ", ".join(choix)
    while True:
        reponse = input(f"Votre réponse ({reponses_possibles}) : ").strip().lower()
        if reponse in choix:
            return reponse
        print(f"Saisie non reconnue. Choisissez parmi : {reponses_possibles}.")


def s_auto_evaluer(moment):
    """Demande à l'élève d'évaluer sa confiance pour chaque compétence."""
    print(f"\n--- Auto-évaluation {moment} le quiz ---")
    print("Pour chaque compétence, choisissez : 1 = pas encore, 2 = presque, 3 = oui.")
    evaluations = {}
    for competence in competences:
        while True:
            reponse = input(f"\n{competence} (1, 2 ou 3) : ").strip()
            if reponse in ("1", "2", "3"):
                evaluations[competence] = int(reponse)
                break
            print("Saisie non reconnue. Entrez 1, 2 ou 3.")
    return evaluations


def donner_retour(reponse, bonne_reponse, explication):
    """Affiche un retour immédiat et indique si la réponse est correcte."""
    if reponse == bonne_reponse:
        print("✅ Correct ! " + explication)
        return True
    print("❌ Ce n’est pas la bonne réponse.")
    print("Correction : " + explication)
    return False


def faire_quiz():
    score = 0
    acquis = {competence: 0 for competence in competences}

    print("\n==============================")
    print("       Début du quiz S1bis")
    print("==============================")

    # Questions 1 à 3 : reconnaître les structures.
    questions_reconnaissance = [
        ("Q1 — notes = [5, 4.5, 6]. Quelle est la structure utilisée ?", "a",
         "Une liste. Les crochets [ ] indiquent ici une liste."),
        ("Q2 — position = (120, 75). Quelle est la structure utilisée ?", "b",
         "Un tuple. Les parenthèses ( ) indiquent ici un tuple."),
        ("Q3 — joueur = {'nom': 'Link', 'vie': 100}. Quelle est la structure utilisée ?", "c",
         "Un dictionnaire. Il associe des clés, comme 'nom', à des valeurs."),
    ]
    choix_structures = {"a": "Liste", "b": "Tuple", "c": "Dictionnaire"}
    for question, bonne_reponse, explication in questions_reconnaissance:
        reponse = demander_choix(question, choix_structures)
        if donner_retour(reponse, bonne_reponse, explication):
            score += 1
            acquis[competences[0]] += 1

    # Question 4 : déballer un tuple.
    reponse = demander_choix(
        "Q4 — On a position = (200, 150). Quelle ligne permet de récupérer les deux valeurs ?",
        {
            "a": "x, y = position",
            "b": "position = x, y",
            "c": "x = position",
        },
    )
    if donner_retour(
        reponse, "a",
        "La bonne ligne est x, y = position. Les deux valeurs du tuple sont placées dans x et y. (2 points)",
    ):
        score += 2
        acquis[competences[1]] += 2

    # Question 5 : ajouter une clé à un dictionnaire.
    reponse = demander_choix(
        "Q5 — Comment ajouter la clé 'niveau' avec la valeur 3 au dictionnaire joueur ?",
        {
            "a": "joueur.update('niveau', 3)",
            "b": "joueur['niveau'] = 3",
            "c": "joueur['niveau': 3]",
        },
    )
    if donner_retour(
        reponse, "b",
        "On écrit joueur['niveau'] = 3. Une clé nouvelle est ainsi ajoutée au dictionnaire. (2 points)",
    ):
        score += 2
        acquis[competences[2]] += 2

    # Question 6 : choisir une structure adaptée.
    reponse = demander_choix(
        "Q6 — Quel choix associe correctement chaque situation à sa structure ?\n"
        "Ordre : liste de notes / coordonnées (x, y) / fiche descriptive d’un personnage",
        {
            "a": "Liste / Tuple / Dictionnaire",
            "b": "Tuple / Liste / Dictionnaire",
            "c": "Dictionnaire / Tuple / Liste",
        },
    )
    if donner_retour(
        reponse, "a",
        "Une liste convient pour une collection de notes, un tuple pour les coordonnées, et un dictionnaire pour des informations nommées. (3 points)",
    ):
        score += 3
        acquis[competences[3]] += 3

    # Question 7 : reconnaître un dictionnaire.
    reponse = demander_choix(
        "Q7 — Quel programme utilise un dictionnaire ?",
        {
            "a": "A : joueur = ('Mario', 100)",
            "b": "B : joueur = {'nom': 'Mario', 'vie': 100}",
        },
    )
    if donner_retour(
        reponse, "b",
        "Le programme B utilise un dictionnaire : les informations sont associées aux clés 'nom' et 'vie'.",
    ):
        score += 1
        acquis[competences[0]] += 1

    return score, acquis


evaluation_avant = s_auto_evaluer("avant")

while True:
    score, acquis = faire_quiz()

    print("\n==============================")
    print(f"Résultat : {score}/11 points")
    print("==============================")
    if score == 11:
        print("Excellent : toutes les réponses sont correctes !")
    elif score >= 8:
        print("Bon travail : révisez les compétences où vous avez perdu des points.")
    elif score >= 5:
        print("Vous progressez : relisez les corrections et entraînez-vous encore.")
    else:
        print("C’est le moment de revoir les exemples du cours, puis de refaire le quiz.")

    print("\nBilan par compétence :")
    maximums = [4, 2, 2, 3]
    for competence, maximum in zip(competences, maximums):
        points = acquis[competence]
        statut = "à réviser" if points == 0 else "en cours" if points < maximum else "réussi"
        print(f"- {competence} : {points}/{maximum} — {statut}")

    evaluation_apres = s_auto_evaluer("après")
    print("\nComparez votre confiance avant et après le quiz :")
    for competence in competences:
        avant = evaluation_avant[competence]
        apres = evaluation_apres[competence]
        evolution = "en hausse" if apres > avant else "en baisse" if apres < avant else "stable"
        print(f"- {competence} : {avant} → {apres} ({evolution})")

    refaire = input("\nVoulez-vous refaire le quiz ? (o/n) : ").strip().lower()
    while refaire not in ("o", "n"):
        refaire = input("Répondez par o ou n : ").strip().lower()
    if refaire == "n":
        print("\nBravo pour votre travail. Pensez à noter les notions à revoir !")
        break
```
